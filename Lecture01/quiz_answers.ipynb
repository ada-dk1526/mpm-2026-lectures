# Lecture 1 self-check: GitHub and Python environments: answers

The answers to the Lecture 1 self-check, with a short explanation for each, including why the tempting wrong options are wrong.

### Question 1

In a freshly cloned repository you add a new function to the tracked file `analysis.py` and also create a new file `utils.py`. You then run

```bash
git commit -a -m "Add max_level"
```

What does `git status` now report for `utils.py`?

- **(A)** It is listed under "Changes to be committed"
- **(B)** It is listed under "Untracked files"
- **(C)** It is listed under "Changes not staged for commit"
- **(D)** Nothing: it was committed along with `analysis.py`

**Answer: (B)**

`-a` automatically stages files that are already tracked and have been modified or deleted, but not newly created files. `analysis.py` is committed; `utils.py` was never added, so it is still untracked. To include it you must `git add utils.py` first. It was not committed (that would need `git add`), it is not staged (nothing staged it), and "Changes not staged for commit" is only for files git already tracks.


### Question 2

A repository on GitHub has 3 commits. You clone it, edit the tracked file `README.md`, and run

```bash
git commit -a -m "Describe the project"
```

You run no other `git` command. How many commits do your local repository and the GitHub repository now have?

- **(A)** Local: 4, GitHub: 3
- **(B)** Local: 4, GitHub: 4
- **(C)** Local: 3, GitHub: 3

**Answer: (A)**

A commit is recorded only in your local repository. GitHub knows nothing about it until you run `git push`, so locally there are 4 commits and GitHub still has 3. Committing does not upload anything. And `-a` does record the change here, because `README.md` is already tracked: it stages modified tracked files automatically.


### Question 3

You clone a repository and, without changing anything else, create a new file `plotting.py` containing a few lines of code, and run

```bash
git add plotting.py
```

Which of the following describes what `git diff` and `git diff --staged` now print?

- **(A)** Both commands show the contents of `plotting.py`
- **(B)** `git diff` shows the contents of `plotting.py`; `git diff --staged` prints nothing
- **(C)** `git diff` prints nothing; `git diff --staged` shows the contents of `plotting.py`
- **(D)** Neither command prints anything

**Answer: (C)**

Plain `git diff` compares your working files with the *staging area*. `git add` has just copied `plotting.py` into the staging area, so the two agree and `git diff` prints nothing. `git diff --staged` compares the staging area with the last commit, i.e. it shows what the next commit will contain, so the new file appears there. Answering that both commands show the file wrongly treats plain `git diff` as showing everything changed since the last commit. The reverse answer swaps the roles of the two commands. And a file does not need to be committed to appear in a diff: once staged, it shows up in `git diff --staged` as a new file.


### Question 4

Two teammates, Alice and Bob, each create a branch (`alice` and `bob`) from the same up-to-date `main`. Alice adds a function `min_level` at the end of `analysis.py`; Bob adds a different function, `tidal_range`, at the end of the same file. They both commit, push their branches and open pull requests, and Alice's pull request is merged into `main`. What happens when someone then tries to merge Bob's pull request?

- **(A)** GitHub merges it automatically, and `analysis.py` on `main` contains both new functions
- **(B)** GitHub merges it automatically, and Bob's version of `analysis.py` replaces Alice's
- **(C)** GitHub reports that the pull request has a conflict in `analysis.py`, which must be resolved first

**Answer: (C)**

Both branches changed the same place, the end of `analysis.py`, relative to the `main` they started from. Git (which GitHub uses to merge pull requests) cannot decide what order the two additions should go in, so the pull request reports a conflict that someone has to resolve (the lecture's team exercise produces exactly this). Git merges lines of text, not functions, so "they are different functions" does not help. And a merge never silently throws away one side's changes: when git cannot combine them, it asks you.


### Question 5

You run `git pull --ff-only` on your local `main` in two different situations.

- **P:** GitHub's `main` has a new commit, and your local `main` has no commits of its own.
- **Q:** GitHub's `main` has a new commit, and you have also committed a change to a *different* file on your local `main`.

What does git do in each?

- **(A)** P: fast-forwards `main`. Q: refuses, leaving `main` unchanged
- **(B)** P: fast-forwards `main`. Q: creates a merge commit on `main`
- **(C)** P: refuses, leaving `main` unchanged. Q: refuses, leaving `main` unchanged
- **(D)** P: creates a merge commit on `main`. Q: creates a merge commit on `main`

**Answer: (A)**

`--ff-only` updates your branch only if it can simply be fast-forwarded to match the remote. In P your local `main` has nothing the remote lacks, so it is fast-forwarded. In Q the two branches have diverged: each has a commit the other lacks. A plain `git pull` would create a merge commit here, but `--ff-only` refuses instead, even though the commits touch different files and would merge without a conflict. It also does not refuse just because GitHub has a commit you don't; that is exactly the case it handles. And a pull that can fast-forward (P) never needs a merge commit.


### Question 6

Project X requires `pandas>=2.2,<3.0` and project Y requires `pandas>=3.0`. The available versions of `pandas` are 2.1.4, 2.2.3, 2.3.1 and 3.0.2. Which of these versions satisfy both projects' requirements, so that a single environment could serve both?

- **(A)** 3.0.2 only
- **(B)** 2.2.3 and 2.3.1
- **(C)** 2.2.3, 2.3.1 and 3.0.2
- **(D)** None of them

**Answer: (D)**

X accepts 2.2.3 and 2.3.1 (at least 2.2, and strictly below 3.0). Y accepts only 3.0.2. No version satisfies both, so one shared environment cannot serve both projects. This is exactly why each project gets its own virtual environment. "2.2.3 and 2.3.1" checks only X's requirement. "3.0.2 only" comes from reading `<3.0` as if it allowed 3.0.x releases. "2.2.3, 2.3.1 and 3.0.2" gives each project a version it accepts, but one environment can only hold one version of `pandas` at a time.


### Question 7

You want to check that your package also works on a different Linux distribution from the one your laptop runs. Which of the tools from the lecture lets you do that?

- **(A)** uv
- **(B)** Docker
- **(C)** conda

**Answer: (B)**

A Docker container encapsulates an entire operating system, so you can run your code on, say, Debian or Alpine while your laptop runs Ubuntu. conda can also provide non-Python libraries and compilers, and uv can even install a different version of Python, but both still run on your host operating system: they change what is installed, not the operating system itself.


### Question 8

A project's `pyproject.toml` declares the package `name = "tidewatch"`, and its `environment.yml` is

```yaml
name: tide-lab
channels:
  - conda-forge
  - nodefaults
dependencies:
  - python=3.14
  - pip
  - numpy
  - pip:
    - -e .
```

After running `conda env create -f environment.yml`, which command activates the new environment?

- **(A)** `source tidewatch/bin/activate`
- **(B)** `source tide-lab/bin/activate`
- **(C)** `conda activate tidewatch`
- **(D)** `conda activate tide-lab`

**Answer: (D)**

The environment's name comes from the `name:` field of `environment.yml`, here `tide-lab`, and a conda environment is activated with `conda activate <name>`. `tidewatch` is the name of the *package* being installed into the environment (set in `pyproject.toml`), a different thing. `source <name>/bin/activate` is how you activate a `venv`, which lives in a folder of that name; conda environments are activated with `conda activate`.


### Question 9

A package `tidewatch` has this `tidewatch/__init__.py`:

```python
from .filters import *  # noqa
```

and this `tidewatch/filters.py`:

```python
__all__ = ['running_mean']


def running_mean(levels, n):
    return [sum(levels[i:i + n]) / n for i in range(len(levels) - n + 1)]


def detrend(levels):
    mean = sum(levels) / len(levels)
    return [h - mean for h in levels]
```

What happens when you run `from tidewatch import running_mean` and then `from tidewatch import detrend`?

- **(A)** `running_mean` raises an `ImportError`; `detrend` raises an `ImportError`
- **(B)** `running_mean` is imported; `detrend` is imported
- **(C)** `running_mean` is imported; `detrend` raises an `ImportError`

**Answer: (C)**

`from .filters import *` copies into the package only the names listed in `filters.py`'s `__all__`. `running_mean` is listed, so `from tidewatch import running_mean` works. `detrend` is not, so it never reaches the package namespace. This is why the lecture's exercise adds each new function to `__all__`: being defined in `filters.py` is not enough. And the star-import does make the listed names available directly from `tidewatch`, so they need not be imported from `tidewatch.filters`.


### Question 10

Running a script gives

```
Traceback (most recent call last):
  File "/home/student/tidewatch/scripts/plot_levels.py", line 1, in <module>
    from tidewatch import smooth_levels
  File "/home/student/tidewatch/tidewatch/__init__.py", line 1, in <module>
    from .filters import *  # noqa
  File "/home/student/tidewatch/tidewatch/filters.py", line 2, in <module>
    from skimage.filters import gaussian
ModuleNotFoundError: No module named 'skimage'
```

Which package should you add to `requirements.txt` (and install) to fix this?

- **(A)** `tidewatch`
- **(B)** `scikit-image`
- **(C)** `skimage`

**Answer: (B)**

Read a traceback from the bottom: the last line names the module that could not be found, `skimage`. That module is provided by the package `scikit-image`, which is what the lecture's own exercise adds to `requirements.txt` for `from skimage import data`. The name you import and the name you install are not always the same. `tidewatch` appears at the top of the traceback only because it is the package whose import triggered the chain; it is the project itself, not what is missing.


### Question 11

You have cloned your fork of a package repository whose committed `.gitignore` contains

```
envtest/__pycache__/
envtest.egg-info/
```

After installing the package with `pip install -e .` and working on it (without running any `git` commands), these three paths have appeared or changed. Which one does `git status` list under "Untracked files"?

- **(A)** `scripts/smooth_image.py`, a script from the repository that you have edited
- **(B)** `envtest/__pycache__/builtins.cpython-314.pyc`, created when the package was imported
- **(C)** `scripts/solve_matrix_equation.py`, a new script you have written

**Answer: (C)**

Untracked means new to git, not yet added, and not ignored. `envtest/__pycache__/` holds the bytecode Python generates when the package is imported, and the `.gitignore` excludes it, so `git status` does not list it at all. `scripts/smooth_image.py` is already tracked, so your edit appears under "Changes not staged for commit", not "Untracked files". Only the new script `scripts/solve_matrix_equation.py` is listed as untracked: it is the one file here that git is not tracking at all.


### Question 12

Following the lecture's recommendations for your own projects, which file should be the single source of truth for your package's runtime dependencies (for example `numpy>=2.1`)?

- **(A)** `pyproject.toml`
- **(B)** `requirements.txt`
- **(C)** `constraints.txt`
- **(D)** `environment.yml`

**Answer: (A)**

The lecture recommends `pyproject.toml` as the single source of truth for runtime dependencies and optional extras. `pip install -e .` reads them from there. `environment.yml` should pin Python and a few heavy libraries for a conda environment, and then install your package from `pyproject.toml`. `requirements.txt` is often not needed, and a pinned one should be generated rather than hand-maintained. A constraints file holds version pins applied with `pip install -c`, not the list of dependencies.
