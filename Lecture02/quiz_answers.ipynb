### Question 1

What arrays do these two lines create?

```python
a = np.arange(0, 1, 0.25)
b = np.linspace(0, 1, 4)
```

- **(A)** `a` is `[0, 0.25, 0.5, 0.75, 1]`; `b` is `[0, 0.25, 0.5, 0.75]`
- **(B)** `a` is `[0, 0.25, 0.5, 0.75, 1]`; `b` is `[0, 1/3, 2/3, 1]`
- **(C)** `a` is `[0, 0.25, 0.5, 0.75]`; `b` is `[0, 1/3, 2/3, 1]`
- **(D)** `a` is `[0, 0.25, 0.5, 0.75]`; `b` is `[0, 0.25, 0.5, 0.75]`

**Answer: (C)**

`np.arange(start, stop, step)` works like `range`: it counts up from `start` in steps of `step` and stops *before* `stop`, so here `a` holds 0, 0.25, 0.5 and 0.75, but not 1. (With a step that is not exactly representable, rounding can occasionally let a value next to `stop` slip in, which is one reason to prefer `linspace` for evenly spaced floats.) `np.linspace(start, stop, num)` gives `num` evenly spaced points with *both* ends included, so four points from 0 to 1 are 1/3 apart: 0, 1/3, 2/3 and 1. The options with a fifth value in `a` include the stop, which `arange` does not do here. Those with quarters in `b` treat `linspace` as leaving out its end point, like `arange`, which it does not.


### Question 2

What is `x.shape`?

```python
x = np.array([[[7, 0, 9], [4, 8, 5]]])
```

- **(A)** `(1, 2, 3)`
- **(B)** `(3, 2, 1)`
- **(C)** `(2, 3, 1)`
- **(D)** `(2, 3)`

**Answer: (A)**

Count the brackets from the outside in. The outermost pair holds one element (the block `[[7, 0, 9], [4, 8, 5]]`), which holds two lists of three numbers each. So the shape is `(1, 2, 3)`, and `x.ndim` is 3. Leaving out the outer level gives `(2, 3)`, but that extra pair of brackets is a dimension of length 1. `(3, 2, 1)` lists the lengths from the inside out, and `(2, 3, 1)` counts the length-1 dimension but puts it last: the outermost brackets always give the first entry of the shape.


### Question 3

What does this print?

```python
a = np.arange(10)
print(a[8:2:-2])
```

- **(A)** `[8 6 4 2]`
- **(B)** `[8 6 4]`
- **(C)** `[]`
- **(D)** `[2 4 6]`

**Answer: (B)**

A slice `start:stop:step` starts at index `start` and moves by `step` until it reaches `stop`. The stop is never included, whichever direction the slice goes. With a step of -2, it visits 8, 6 and 4 and stops before index 2, so it prints `[8 6 4]` (here each element equals its index). Including 2 treats the stop as inclusive, and `[2 4 6]` swaps the start and stop and counts upwards, ignoring the sign of the step. A negative step also does not need `start < stop`: when counting down, the start must be the larger index, as it is here, so the slice is not empty.


### Question 4

What does this print?

```python
a = np.array([1, 2, 3])
b = a
c = a.copy()
a[0] = 99
print(b[0], c[0])
```

- **(A)** `99 1`
- **(B)** `1 1`
- **(C)** `1 99`
- **(D)** `99 99`

**Answer: (A)**

`b = a` does not copy the array: it gives the same array a second name. So a change made through `a` is seen through `b`, and `b[0]` is 99. `a.copy()` makes an independent array with its own copy of the data, taken before the change, so `c[0]` is still 1. The other options assume that assignment copies the array (`1 1`), that `copy()` still shares the data (`99 99`), or that the two behave the other way round (`1 99`).


### Question 5

Each of the snippets (i), (ii) and (iii) below starts from a fresh

```python
p = np.arange(12).reshape(3, 4)
```

```python
# (i)
q = p[[0, 2], 1]
q[0] = -1

# (ii)
s = p[::2, 1]
s[0] = -1

# (iii)
r = p.reshape(4, 3)
r[0, 0] = -1
```

Which of them change `p`?

- **(A)** (i) and (ii) only
- **(B)** None of them
- **(C)** (i) and (iii) only
- **(D)** (ii) and (iii) only

**Answer: (D)**

(i) and (ii) pick out the same two elements, `p[0, 1]` and `p[2, 1]`, in two different ways. Indexing with a list of indices (fancy indexing, (i)) returns a *copy*, so `q` has its own data and changing it leaves `p` alone. A slice (ii) returns a *view* of `p`'s data, so `s[0] = -1` sets `p[0, 1]` to -1. `reshape` (iii) also returns a view whenever it can, as it can here, so `r[0, 0] = -1` changes `p[0, 0]`. So (ii) and (iii) change `p`. Binding the result to a new name does not by itself make it independent of `p`: what matters is whether the operation made a view or a copy.


### Question 6

With

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[0, 1], [1, 0]])
```

what are `A * B` and `A @ B`?

- **(A)** Both are `[[2, 1], [4, 3]]`
- **(B)** Both are `[[0, 2], [3, 0]]`
- **(C)** `A * B` is `[[2, 1], [4, 3]]`; `A @ B` is `[[0, 2], [3, 0]]`
- **(D)** `A * B` is `[[0, 2], [3, 0]]`; `A @ B` is `[[2, 1], [4, 3]]`

**Answer: (D)**

For NumPy arrays, `*` multiplies element by element: each entry of `A` is multiplied by the entry in the same place in `B`, giving `[[1*0, 2*1], [3*1, 4*0]]`, i.e. `[[0, 2], [3, 0]]`. The matrix product is `@` (or `np.dot`): entry (i, j) combines row i of `A` with column j of `B`. For example, the top-left entry is `1*0 + 2*1 = 2`, and the result is `[[2, 1], [4, 3]]` (this `B` swaps `A`'s columns). The two operations give different results, and for NumPy arrays `*` is never the matrix product.


### Question 7

With

```python
a = np.array([[1, 2, 3], [4, 5, 6]])
```

what are `a.T` and `a.reshape(3, 2)`?

- **(A)** Both are `[[1, 2], [3, 4], [5, 6]]`
- **(B)** `a.T` is `[[1, 4], [2, 5], [3, 6]]`; `a.reshape(3, 2)` is `[[1, 2], [3, 4], [5, 6]]`
- **(C)** `a.T` is `[[1, 2], [3, 4], [5, 6]]`; `a.reshape(3, 2)` is `[[1, 4], [2, 5], [3, 6]]`
- **(D)** Both are `[[1, 4], [2, 5], [3, 6]]`

**Answer: (B)**

Both results have shape `(3, 2)`, but they are different arrays. The transpose `a.T` turns rows into columns: its first row is `a`'s first column, `[1, 4]`. `reshape` never reorders the elements. It reads them in their existing order, 1, 2, 3, 4, 5, 6 (row by row), and fills the new shape in the same order, giving `[[1, 2], [3, 4], [5, 6]]`. So a transpose is not a reshape to the swapped shape, and treating them as the same is a common mistake.


### Question 8

What does this print?

```python
x = np.array([100, 200], dtype=np.uint8)
print(x + 100)
```

- **(A)** `[200  44]`
- **(B)** `[200 300]`
- **(C)** Nothing: it raises an `OverflowError`
- **(D)** `[200 255]`

**Answer: (A)**

`uint8` elements are 8-bit unsigned integers, which can only hold values from 0 to 255. The result keeps the array's data type, so 200 + 100 = 300 does not fit and silently wraps round to 300 - 256 = 44. NumPy does not switch to a bigger integer type to make room (`[200 300]`). It does not stop at the largest value (`[200 255]`), and it does not raise an error for this arithmetic on arrays. Choosing a data type big enough for your results is up to you.


### Question 9

What does this print?

```python
m = np.array([[0, -5, 4], [-1, 2, -3]])
print(m[m < 0])
```

- **(A)** `[-1 -5 -3]`
- **(B)** `[-5 -3 -1]`
- **(C)** `[ 0 -1 -5 -3]`
- **(D)** `[-5 -1 -3]`

**Answer: (D)**

`m < 0` is a boolean array with the same shape as `m`, `True` where an element is negative. Indexing with it, `m[m < 0]`, selects just those elements and returns them as a 1D array, in the order they appear in `m`, row by row: -5 from the first row, then -1 and -3 from the second, so it prints `[-5 -1 -3]`. Reading down the columns instead gives `[-1 -5 -3]`, and the elements are not sorted (`[-5 -3 -1]`). And the 0 is not selected: `0 < 0` is `False`, so including it treats `<` as `<=`.


### Question 10

Each option below defines `rng1` and `rng2`. Then

```python
x = rng1.random(3)
y = rng2.random(3)
```

runs. With which set-up do `x` and `y` hold exactly the same numbers?

- **(A)** `rng1 = np.random.default_rng()` and `rng2 = np.random.default_rng()`
- **(B)** `np.random.seed(42)`, then `rng1 = np.random.default_rng()` and `rng2 = np.random.default_rng()`
- **(C)** `rng1 = np.random.default_rng(42)` and `rng2 = np.random.default_rng(42)`
- **(D)** `rng1 = np.random.default_rng(42)` and `rng2 = rng1`

**Answer: (C)**

A generator's numbers are completely determined by its seed. Two generators created with the same seed produce exactly the same sequence, which is how you make a result reproducible. Without a seed, `default_rng()` takes fresh entropy from the operating system, so two unseeded generators give different numbers (as does the same notebook run twice). `rng2 = rng1` does not make a second generator: both names refer to one generator, so `y` is the *next* three numbers in its sequence, not the same ones again. And `np.random.seed(42)` seeds only NumPy's legacy global generator (used by `np.random.rand` and friends); it has no effect on generators made with `default_rng()`.


### Question 11

With

```python
A = np.array([[2, 1], [4, 3]])
b = np.array([3, 5])
```

which line computes the vector `x` for which `A @ x` equals `b`?

- **(A)** `x = b @ np.linalg.inv(A)`
- **(B)** `x = b / A`
- **(C)** `x = np.linalg.solve(A, b)`
- **(D)** `x = np.linalg.inv(A) * b`

**Answer: (C)**

`np.linalg.solve(A, b)` solves the linear system `A @ x == b` directly, giving x = [2, -1] (printed as `[ 2. -1.]`; check: `A @ x` gives `b`). `b @ np.linalg.inv(A)` multiplies by the inverse on the wrong side: it solves `x @ A == b` instead, which gives a different answer here because `A` is not symmetric. `np.linalg.inv(A) * b` multiplies the inverse by `b` *element by element* (`b` is broadcast across the rows) instead of forming the matrix product, so it gives a 2x2 array, not `x`. `np.linalg.inv(A) @ b` would work, but as the lecture points out, computing the inverse is more expensive than calling `solve`. `b / A` divides element by element, which is not how you 'divide' by a matrix.


### Question 12

After

```python
x = np.linspace(0, 1, 5)
fig, axes = plt.subplots(2, 3)
```

which line draws a curve in the bottom-left panel of the figure?

- **(A)** `axes[2, 1].plot(x, x**2)`
- **(B)** `axes[1, 0].plot(x, x**2)`
- **(C)** `axes[0, 1].plot(x, x**2)`
- **(D)** `axes[1, 2].plot(x, x**2)`

**Answer: (B)**

`plt.subplots(2, 3)` returns the figure and a 2x3 NumPy array of axes, one per panel. The array is indexed `axes[row, column]`, counting from 0, with row 0 at the top. The bottom-left panel is in row 1 (the second and last row) and column 0, so it is `axes[1, 0]`. `axes[0, 1]` puts the column first and draws in the second panel of the top row. `axes[2, 1]` counts from 1, so it raises an `IndexError` (there is no row 2). `axes[1, 2]` both puts the column first and counts from 1, and draws in the bottom-right panel instead.
